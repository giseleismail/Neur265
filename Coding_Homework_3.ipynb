```python
#import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
#import v_rest_sst and v_rest_pvalb
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
v_rest_sst = np.loadtxt(url1)
v_rest_pvalb = np.loadtxt(url2)
```

```python
#means of v_rest_sst and v_rest_pvalb
mean_sst = np.mean(v_rest_sst)
mean_pvalb = np.mean(v_rest_pvalb)

print("Mean sst:", mean_sst)
print("Mean pvalb:", mean_pvalb)
```

```python
#ttest for null hypothesis that means from data come from different distributions

t_stat, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
#plot two histograms in subplots
if p_value < 0.05:
    fig, ax = plt.subplots(1, 2, figsize=(15,4))

    # First histogram
    ax[0].hist(v_rest_sst, color='skyblue')
    ax[0].set_xlabel('Membrane Potential')
    ax[0].set_ylabel('Count')

    # Second histogram
    ax[1].hist(v_rest_pvalb, color='limegreen')
    ax[1].set_xlabel('Resting membrane potential')
    ax[1].set_ylabel('Count')

    plt.show()
else:
    print('p-value >= 0.05, no histogram.')
```

```python
#define vrestHist
def vrestHist():
    mean_sst = np.mean(v_rest_sst)
    mean_pvalb = np.mean(v_rest_pvalb)

    print("Mean sst:", mean_sst)
    print("Mean pvalb:", mean_pvalb)

    t_stat, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb)
    print("t-statistic:", t_stat)
    print("p-value:", p_value)

    if p_value < 0.05:
        fig, ax = plt.subplots(1, 2, figsize=(10,4))

        # First histogram
        ax[0].hist(v_rest_sst, color='skyblue')
        ax[0].set_xlabel('Membrane Potential')
        ax[0].set_ylabel('Count')

        # Second histogram
        ax[1].hist(v_rest_pvalb, color='limegreen')
        ax[1].set_xlabel('Resting Membrane Potential')
        ax[1].set_ylabel('Count')

        plt.tight_layout()
        plt.show()
    else:
        print('p-value >= 0.05, no histogram.')
```

```python
#call vrestHist
vrestHist()
```

```python
# scatter plot with color code
for i in range(len(v_rest_sst)):
  if v_rest_sst[i] > -60 and v_rest_pvalb[i] >-60:
    color = 'red'
  elif v_rest_sst[i] > -60 and v_rest_pvalb[i] <=60:
    color = 'green'
  elif v_rest_sst[i] <= -60 and v_rest_pvalb[i] <=-60:
    color = 'blue'
  elif v_rest_sst[i] <= -60 and v_rest_pvalb[i] > -60:
    color = 'purple'

  plt.scatter(v_rest_sst[i], v_rest_pvalb[i], color=color)

plt.xlabel('v_rest_sst')
plt.ylabel('v_rest_pvalb')
```
