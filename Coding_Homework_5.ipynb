```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
spikes_left_1 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv', delimiter = ',')
spikes_left_2 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv', delimiter = ',')
spikes_right_1 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv', delimiter = ',')
spikes_right_2 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv', delimiter = ',')
```

```python
#find the length of each trial
print("left first trial:", len(spikes_left_1))
print("left second trial:", len(spikes_left_2))
print("right first trial:", len(spikes_right_1))
print("right second trial:", len(spikes_right_2))

```

```python
#find firing rate of neuron during each trial
FR_L1 = len(spikes_left_1)/30
print("Firing rate trial 1 Left =", FR_L1)
FR_L2 = len(spikes_left_2)/30
print("Firing rate trial 2 Left =", FR_L2)
FR_R1 = len(spikes_right_1)/30
print("Firing rate trial 1 Right =", FR_R1)
FR_R2 = len(spikes_right_2)/30
print("Firing rate trial 2 Right =", FR_R2)
```

It looks pretty similar to me, but fires *slightly* more before left turns.

```python
#subplots of spike train data from trial 1 and trial 2

fig, ax = plt.subplots(1,2)
ax[0].scatter(spikes_left_1, [1]*len(spikes_left_1), color = 'red')
ax[0].scatter(spikes_right_1, [0]*len(spikes_right_1), color = 'limegreen')
ax[0].set_xlabel("Time (s)")
ax[0].set_ylabel("Trial Type")
ax[0].set_title('Trial 1')


ax[1].scatter(spikes_left_2, [1]*len(spikes_left_2), color = 'red')
ax[1].scatter(spikes_right_2, [0]*len(spikes_right_2), color = 'limegreen')
ax[1].set_xlabel("Time (s)")
ax[1].set_ylabel("Trial Type")
ax[1].set_title('Trial 2')
```

both left and right look bursty

```python
#ISI for each spike

ISI_left_1 = np.diff(spikes_left_1)
ISI_right_1 = np.diff(spikes_right_1)
ISI_left_2 = np.diff(spikes_left_2)
ISI_right_2 = np.diff(spikes_right_2)
```

```python
# four histograms
bins = np.arange(0, 2, 0.01)

fig, ax = plt.subplots(2, 2, figsize = (10,11))

ax[0,0].hist(ISI_left_1, bins=bins)
ax[0,0].set_title("ISI Left Trial 1")
ax[0,0].set_xlabel("ISI")
ax[0,0].set_ylabel("Count")

ax[0,1].hist(ISI_right_1, bins=bins)
ax[0,1].set_title("ISI Right Trial 1")
ax[0,1].set_xlabel("ISI")
ax[0,1].set_ylabel("Count")

ax[1,0].hist(ISI_left_2, bins=bins)
ax[1,0].set_title("ISI Left Trial 2")
ax[1,0].set_xlabel("ISI")
ax[1,0].set_ylabel("Count")

ax[1,1].hist(ISI_right_2, bins=bins)
ax[1,1].set_title("ISI Right Trial 2")
ax[1,1].set_xlabel("ISI")
ax[1,1].set_ylabel("Count")

plt.show()
```

this neuron does typically fire in bursts it seems to have a higher burstiness for the right trials.

```python
#created new bin variable and got firing rate over time histograms

time_bins = np.arange(0, 31, 1)

FR_L1 = np.histogram(spikes_left_1, time_bins)[0]
FR_R1 = np.histogram(spikes_right_1, time_bins)[0]
FR_L2 = np.histogram(spikes_left_2, time_bins)[0]
FR_R2 = np.histogram(spikes_right_2, time_bins)[0]

fig, ax = plt.subplots(2, 1, figsize=(10,8))
ax[0].plot(time_bins[:-1] + 0.5, FR_L1, color = 'red', label = 'Left Turn')
ax[0].plot(time_bins[:-1] + 0.5, FR_R1, color = 'blue', label = 'Right Turn')
ax[0].set_title('Firing Rate Across Time (T1)')
ax[0].set_xlabel('Time (s)')
ax[0].set_ylabel('Firing Rate')

ax[1].plot(time_bins[:-1] + 0.5, FR_L2, color = 'red', label = 'Left Turn')
ax[1].plot(time_bins[:-1] + 0.5, FR_R2, color = 'blue', label = 'Right Turn')
ax[1].set_title('Firing Rate Across Time (T1)')
ax[1].set_xlabel('Time (s)')
ax[1].set_ylabel('Firing Rate')




```

The firing rate is different at specific timepoints between left and right turn trials. for example, in trial 1, there are two large spikes for left turn around 12 and 16 seconds, but not for right turn.  

The pattern is also different between trial one and trial 2.
