```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#load data
fc_hits = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/fc_hits.csv', delimiter = ',')
fc_mistakes = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/fc_mistakes.csv', delimiter = ',')
ic_hits = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/ic_hits.csv', delimiter = ',')
ic_mistakes = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/ic_mistakes.csv', delimiter = ',')
```

```python
#create a time variable
print(fc_hits.shape)
time = np.linspace(-2, 2, 8001)
```

```python
fig, ax = plt.subplots(1,2, figsize=(15,5))

#ERP fc_hits, color = blue, ERP fc_mistakes, color = red
ax[0].plot(time, fc_hits.mean(0), label='hits', color = 'blue')
ax[0].plot(time, fc_mistakes.mean(0), label='mistakes', color = 'red')
ax[0].set_xlabel('Time [s]')
ax[0].set_ylabel('Voltage [V]')
ax[0].set_title('ERP Frontal Cortex')

# vertical line at time when mouse touched the screen on the x-axis
ax[0].axvline(0, color='black')

#horizontal line at zero point on the y-axis
ax[0].hlines(0, time[0], time[-1], color = 'black')



#ERP ic_hits, color = blue, ERP ic_mistakes, color = red
ax[1].plot(time, ic_hits.mean(0), label='hits', color = 'blue')
ax[1].plot(time, ic_mistakes.mean(0), label='mistakes', color = 'red')
ax[1].set_xlabel('Time [s]')
ax[1].set_ylabel('Voltage [V]')
ax[1].set_title('ERP Insular Cortex')


# vertical line at time when mouse touched the screen on the x-axis
ax[1].axvline(0, color='black')

#horizontal line at zero point on the y-axis
ax[1].hlines(0, time[0], time[-1], color = 'black')
```

It looks like the ERP is larger during hits. 0 value allows you to asses how big your ERP is. closer to the line = flatter = less consistent. It looks like mistakes is a lot closer to the 0 crossing line and the hits has some bigger spikes around 0.5 and 1.5 seconds.
